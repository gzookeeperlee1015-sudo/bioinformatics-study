```python
from pathlib import Path
import gzip

DATA = Path("/mnt/HDD3/class_2025_2/BI_sample_datasets/Data_RNA_seq")

R1 = DATA / "RNAseq_reads_chr17_1.fq.gz"
R2 = DATA / "RNAseq_reads_chr17_2.fq.gz"

print("R1 존재:", R1.exists())
print("R2 존재:", R2.exists())
```

```python
with gzip.open(R1, "rt") as f:
    for read_number in range(1, 5):
        header = f.readline().rstrip()
        sequence = f.readline().rstrip()
        separator = f.readline().rstrip()
        quality = f.readline().rstrip()

        print(f"===== Read {read_number} =====")
        print("이름:", header)
        print("염기서열:", sequence)
        print("구분선:", separator)
        print("품질점수:", quality)
        print("염기서열 길이:", len(sequence))
        print("품질점수 길이:", len(quality))
        print()
```

```python
with gzip.open(R1, "rt") as f1, gzip.open(R2, "rt") as f2:
    for pair_number in range(1, 6):
        r1_record = [f1.readline().rstrip() for _ in range(4)]
        r2_record = [f2.readline().rstrip() for _ in range(4)]

        r1_name = r1_record[0]
        r2_name = r2_record[0]

        print(f"===== Pair {pair_number} =====")
        print("R1 이름:", r1_name)
        print("R2 이름:", r2_name)
        print("이름 일치:", r1_name == r2_name)
        print("R1 길이:", len(r1_record[1]))
        print("R2 길이:", len(r2_record[1]))
        print()
```

```python
def summarize_fastq(fastq_path, max_reads=10000):
    read_count = 0
    lengths = []
    gc_ratios = []
    mean_qualities = []

    with gzip.open(fastq_path, "rt") as f:
        while read_count < max_reads:
            header = f.readline().rstrip()

            if not header:       # 파일 끝
                break

            sequence = f.readline().rstrip()
            separator = f.readline().rstrip()
            quality = f.readline().rstrip()

            # Phred+33 방식으로 품질 문자를 숫자로 변환
            quality_scores = [ord(char) - 33 for char in quality]

            lengths.append(len(sequence))

            gc_count = sequence.count("G") + sequence.count("C")
            gc_ratios.append(gc_count / len(sequence) * 100)

            mean_qualities.append(
                sum(quality_scores) / len(quality_scores)
            )

            read_count += 1

    print("확인한 read 수:", read_count)
    print("평균 read 길이:", sum(lengths) / len(lengths))
    print("평균 GC 비율:", sum(gc_ratios) / len(gc_ratios))
    print("평균 품질점수:", sum(mean_qualities) / len(mean_qualities))
```

```python
print("===== R1 요약 =====")
summarize_fastq(R1)

print("\n===== R2 요약 =====")
summarize_fastq(R2)
```
